As you may have noticed, many of the equations in this course are displayed beautifully—just like in a math textbook. In Lesson 0, we learned that this rendering is done using $\mathrm{\LaTeX}$—and yes, even this nicely formatted LaTeX example is itself produced with LaTeX!

[**LaTeX**](https://en.wikipedia.org/wiki/LaTeX) is a markup language for typesetting mathematical formulas and, along with its predecessor [**TeX**](https://en.wikipedia.org/wiki/TeX), traces its origins to the late 1970s when computer typesetting was still crude and inadequate for high-quality scientific publishing. In 1978, Donald Knuth created $\mathrm{\TeX}$ in response to the declining print quality of his _Art of Computer Programming_ volumes, designing it as a precise, programmable system capable of producing professional mathematical layout. In the early 1980s, Leslie Lamport introduced LaTeX as a layer on top of TeX, providing higher-level commands and structured markup so authors could focus on writing rather than formatting. Together, TeX and LaTeX have since become essential tools for scientific and mathematical communication.

So much for the history... In this lesson, we'll learn some basic LaTeX syntax for use in Markdown cells, and you'll also see how it can be applied in plotting in Lesson 3.

## Prologue: Inline and Displayed Formulae

Before we introduce any LaTeX syntax, we need to know how to display formulas in Markdown. A simple case is when we want to show a formula directly within the text, like $E = mc^2$. This is called an **inline formula**, and it's created by surrounding the expression with a single dollar sign `$` on each side. For example:

```markdown
We would like to know the energy of an electron in a vacuum, $E = mc^2$.
```

Try to type this in the <u>Markdown</u> cell below and see what happens.

Sometimes we want to display a formula on its own line, like this:

$$E = mc^2$$

This is called a **displayed formula**, and it's done using **two** dollar signs `$$` to surround the formula. Try the following:

```markdown
The integral of $f(x)$ from $a$ to $b$ is:
$$\int_a^bf(x)dx$$
```


## Basic Mathematics

### Formulae

All LaTeX codes (wrapped either in single or double dollar signs) in Markdown are rendered as mathematical formulas. This means that all letter symbols are italic, just as you would expect from a textbook. It is really straightforward to typeset some formulas, like this:

```markdown
$$ f(x) = 2x + 1 $$
```

Note that spacing in LaTeX is not important, so you can write the above formula as:

```markdown
$$f(x)=2x+1$$
$$ f ( x ) = 2 x + 1 $$
$$ f(x)= 2     x  + 1$$
```

They should all render the same.

### Superscripts and Subscripts

You can also use superscripts and subscripts in LaTeX. Superscripts are denoted by `^` and subscripts by `_`. For example:

```markdown
$$ x^2 + y^2 = z^2 $$
$$ x = c_1x_1 + c_2x_2 + c_3x_3 $$
```

You may notice that `^` and `_` apply superscripts or subscripts to only the next single character, not to everything that follows. This is important to keep in mind. If you want to apply a superscript or subscript to multiple characters, you must wrap them in braces `{}`:

```markdown
$$ e^{x - y} $$
```

You can also use multiple superscripts and subscripts in a single formula, but you must use braces appropriately:

```markdown
$$ e^{x^2} $$
```
instead of
```markdown
$$ e^x^2 $$
```
